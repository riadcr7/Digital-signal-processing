```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
signal1 = np.loadtxt("/content/Signal1.txt")
signal2 = np.loadtxt("/content/Signal2.txt")
signal3 = np.loadtxt("/content/Signal3.txt")
```

```python
signal1_by5 = signal1.copy()
signal1_by5[:,1] = signal1_by5[:,1] * 5
signals1 = [
        [signal1,"signal1"],
        [signal1_by5,"signal1_by5"]
    ]
for signal , name in signals1:
    plt.figure()
    x = signal[:,0]
    y = signal[:,1]
    plt.plot(x,y)
    plt.title(name)
    plt.grid(True)
    plt.show()
```

```python
signal2_by10 = signal2.copy()
signal2_by10[:,1] = signal2_by10[:,1] * 10
signals2 = [
        [signal2,"signal2"],
        [signal2_by10,"signal2_by10"]
    ]
for signal , name in signals2:
    plt.figure()
    x = signal[:,0]
    y = signal[:,1]
    plt.plot(x,y)
    plt.title(name)
    plt.grid(True)
    plt.show()
```

```python
result = signal1.copy()
result[:,1] += signal2[:,1]

signals3 = [
        [signal1,"signal1"],
        [signal2,"signal2"],
        [result,"result"]
    ]
for signal , name in signals3:
    plt.figure()
    x = signal[:,0]
    y = signal[:,1]
    plt.plot(x,y)
    plt.title(name)
    plt.grid(True)
    plt.show()
```

```python
result = signal1.copy()
result[:,1] += signal3[:,1]

signals4 = [
        [signal1,"signal1"],
        [signal3,"signal3"],
        [result,"result"]
    ]
for signal , name in signals4:
    plt.figure()
    x = signal[:,0]
    y = signal[:,1]
    plt.plot(x,y)
    plt.title(name)
    plt.grid(True)
    plt.show()
```
